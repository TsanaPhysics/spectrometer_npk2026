โค้ดแสดงข้อมูลจากไฟล์ทั้งหมดในโฟลเดอร์ "raw_data" ที่มีนามสกุล ".txt" โดยแสดงข้อมูลของแต่ละไลน์ในแต่ละไฟล์ และเพิ่มบรรทัดว่างหลังจากแต่ละไฟล์เพื่อให้ข้อมูลดูเรียบร้อยขึ้นในผลลัพธ์ที่แสดงออกมาแบบเป็นรายการตามไฟล์ที่อยู่ในโฟลเดอร์ "raw_data" ทั้งหมด.

```python
import os

# ระบุโฟลเดอร์ raw_data
folder_path = "raw_data"

# ตรวจสอบว่าโฟลเดอร์นี้มีอยู่จริงหรือไม่
if os.path.exists(folder_path):
    # รายการไฟล์ในโฟลเดอร์
    files = os.listdir(folder_path)

    # วนลูปผ่านไฟล์ทั้งหมดในโฟลเดอร์
    for filename in files:
        file_path = os.path.join(folder_path, filename)

        # ตรวจสอบว่าไฟล์นี้เป็นไฟล์ข้อความหรือไม่
        if filename.endswith(".txt"):
            print(f"แสดงข้อมูลจากไฟล์ '{filename}':")
            
            # เปิดไฟล์สำหรับอ่าน
            with open(file_path, "r") as text_file:
                # อ่านและแสดงข้อมูลของแต่ละไลน์
                for line in text_file:
                    print(line.strip())  # ใช้ strip() เพื่อลบช่องว่างที่อาจจะเกิดขึ้น

            print("\n")  # เพิ่มบรรทัดว่างหลังจากแสดงข้อมูลของแต่ละไฮน์
else:
    print(f"โฟลเดอร์ '{folder_path}' ไม่มีอยู่หรือไม่สามารถเข้าถึงได้")
```

โค้ดแปลงไฟล์ .txt ทุกไฟล์ในโฟลเดอร์ "raw_data" เป็นไฟล์ .csv และบันทึกในโฟลเดอร์ "datasetCSV" โดยลบข้อมูลในแถวแรกของแต่ละไฟล์ .txt ก่อนจะเขียนลงในไฟล์ .csv ที่สร้างขึ้นใหม่

```python
import os
import csv

# ระบุโฟลเดอร์ raw_data
input_folder_path = "raw_data"

# ระบุโฟลเดอร์ที่จะบันทึกไฟล์ .csv
output_folder_path = "datasetCSV"

# ตรวจสอบว่าโฟลเดอร์ output_folder_path มีอยู่หรือไม่ และสร้างถ้าไม่มี
if not os.path.exists(output_folder_path):
    os.makedirs(output_folder_path)

# รายการไฟล์ในโฟลเดอร์ input_folder_path
files = os.listdir(input_folder_path)

# วนลูปผ่านไฟล์ทั้งหมดในโฟลเดอร์ input_folder_path
for filename in files:
    input_file_path = os.path.join(input_folder_path, filename)

    # ตรวจสอบว่าไฟล์นี้เป็นไฟล์ข้อความหรือไม่
    if filename.endswith(".txt"):
        # สร้างชื่อไฟล์ .csv ใหม่โดยนำชื่อไฟล์เดิมมาใช้
        output_file_path = os.path.join(output_folder_path, os.path.splitext(filename)[0] + ".csv")

        # เปิดไฟล์ .txt สำหรับอ่าน
        with open(input_file_path, "r") as txt_file:
            # อ่านข้อมูลทั้งหมดจากไฟล์ .txt
            lines = txt_file.readlines()

        # ตรวจสอบว่ามีข้อมูลในไฟล์หรือไม่
        if len(lines) > 0:
            # ลบข้อมูลในแถวแรก
            lines = lines[1:]

            # เปิดไฟล์ .csv สำหรับเขียน
            with open(output_file_path, "w", newline='') as csv_file:
                # สร้างอ็อบเจ็กต์ writer สำหรับเขียนข้อมูล CSV
                csv_writer = csv.writer(csv_file)

                # เขียนข้อมูลลงในไฟล์ .csv
                for line in lines:
                    # แบ่งข้อมูลในแต่ละบรรทัดด้วยเครื่องหมาย "," และลบช่องว่างด้านหน้าและหลัง
                    row_data = line.strip().split(",")
                    csv_writer.writerow(row_data)

        print(f"แปลงไฟล์ '{filename}' และบันทึกเป็น '{os.path.basename(output_file_path)}'")

print("เสร็จสิ้นการแปลงไฟล์ .txt เป็น .csv และบันทึกในโฟลเดอร์ datasetCSV")
```

# แสดงค่าเฉลี่ยและส่วนเบี่ยงเบนมาตรฐานของแต่ละ Order ของแต่ละ Color  ของแต่ละไฟล์

```python
import os
import pandas as pd
import numpy as np

# Specify the folder path
folder_path = "datasetCSV"

# List all CSV files in the folder
csv_files = [f for f in os.listdir(folder_path) if f.endswith(".csv")]

# Create empty lists to store the calculated statistics
average_intensity_values = []
std_dev_intensity_values = []

# Loop through each CSV file and calculate statistics for 'Intensity' column
for csv_file in csv_files:
    # Construct the full file path
    file_path = os.path.join(folder_path, csv_file)

    # Read the CSV file into a DataFrame
    df = pd.read_csv(file_path)

    # Group data by 'Order' and 'Color' columns, then calculate mean and standard deviation
    grouped_data = df.groupby(['Order', 'Color'])['Intensity']
    mean_intensity = grouped_data.mean()
    std_dev_intensity = grouped_data.std()

    # Append the calculated statistics to the lists
    average_intensity_values.append(mean_intensity)
    std_dev_intensity_values.append(std_dev_intensity)

# Combine the calculated statistics from all files into one DataFrame
average_intensity_df = pd.concat(average_intensity_values, axis=1)
std_dev_intensity_df = pd.concat(std_dev_intensity_values, axis=1)

# Rename columns to indicate file names
average_intensity_df.columns = csv_files
std_dev_intensity_df.columns = csv_files

# Print the calculated statistics
print("Average Intensity Values:")
print(average_intensity_df)

print("\nStandard Deviation of Intensity Values:")
print(std_dev_intensity_df)
```

```python
import os
import pandas as pd

# Define the folder path
folder_path = "datasetCSV"

# List all CSV files in the folder
csv_files = [f for f in os.listdir(folder_path) if f.startswith("MSD_") and f.endswith(".csv")]

for csv_file in csv_files:
    # Construct the full file path
    original_file_path = os.path.join(folder_path, csv_file)

    # Read the CSV file into a DataFrame
    grouped_df = pd.read_csv(original_file_path)

    # Create a pivot table with 'Order' as columns and 'Color' as index
    pivot_df = grouped_df.pivot_table(values='Intensity', index='Color', columns='Order', aggfunc='mean')

    # Calculate the standard deviation for each row (Color)
    std_deviation = pivot_df.apply(lambda row: row.std(), axis=1)

    # Add the mean and standard deviation columns to the pivot table
    pivot_df['Mean'] = pivot_df.mean(axis=1)
    pivot_df['Std Deviation'] = std_deviation

    # Print the pivot table
    print(f"Pivot Table for {csv_file}:")
    print(pivot_df)
    print("\n")

    # Save the pivot table to a CSV file with the same name in the same folder
    output_file_path = os.path.join(folder_path, csv_file)
    pivot_df.to_csv(output_file_path, index=True)
```

```python
import os
import pandas as pd

# ระบุคอลัมน์ที่ต้องการแสดง
columns_to_display = ["Color", "Mean", "Std Deviation"]

# ระบุโฟลเดอร์ที่เก็บไฟล์ CSV
folder_path = 'datasetCSV'

# หาไฟล์ CSV ทั้งหมดในโฟลเดอร์
csv_files = [f for f in os.listdir(folder_path) if f.startswith('MSD_P') and f.endswith('.csv')]

# สร้างรายการของ DataFrame ที่จะรวมข้อมูล
dataframes_to_concat = []

# วนลูปเพื่ออ่านและรวมข้อมูลจากไฟล์ CSV แต่ละไฟล์
for csv_file in csv_files:
    file_path = os.path.join(folder_path, csv_file)
    
    # ใช้ pandas อ่านข้อมูลจากไฟล์ CSV
    df = pd.read_csv(file_path)
    
    # เลือกข้อมูลเฉพาะคอลัมน์ที่ต้องการแสดง
    selected_columns = df[columns_to_display]
    
    # เพิ่ม DataFrame ที่ได้สร้างลงในรายการ
    dataframes_to_concat.append(selected_columns)

# รวมข้อมูลในรายการ DataFrame โดยใช้ pd.concat()
all_data = pd.concat(dataframes_to_concat, ignore_index=True)

# บันทึก DataFrame ที่รวมข้อมูลลงในไฟล์ "AllMSD.csv"
output_file_path = os.path.join(folder_path, 'AllMSD.csv')
all_data.to_csv(output_file_path, index=False)

print(f"บันทึกข้อมูลลงในไฟล์ {output_file_path}")
```
