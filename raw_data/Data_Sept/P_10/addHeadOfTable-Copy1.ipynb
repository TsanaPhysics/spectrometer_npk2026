```python
import pandas as pd

# Define the column names
column_names = ["No.", "Order", "Color", "Value"]

# Load the CSV file with the specified column names
file_path = "datasets/dataHead.csv"
df = pd.read_csv(file_path, names=column_names)

# Print specific columns
columns_to_print = ["No.", "Order", "Color", "Value"]
print(df[columns_to_print])

# Specify the new file path and save the modified DataFrame
new_output_file_path = "datasets/dataHead.csv"  # Change this path as needed
df.to_csv(new_output_file_path, index=False)

# Print a confirmation message
print(f"DataFrame saved to {new_output_file_path}")
```

```python
import pandas as pd

# Define the column names
column_names = ["No.", "Order", "Color", "Value"]

# Load the CSV file with the specified column names
file_path = "datasets/dataHead.csv"
df = pd.read_csv(file_path, names=column_names)

# Filter rows where Color is "Red" and Order is 1
filtered_df = df[(df['Color'] == 'Red') & (df['Order'] == '1')]

# Print the filtered DataFrame
print(filtered_df)
```

```python
import pandas as pd

# Define the column names
column_names = ["No.", "Order", "Color", "Value"]

# Load the CSV file with the specified column names
file_path = "datasets/dataHead.csv"
df = pd.read_csv(file_path, names=column_names)

# Sort the DataFrame by the "Order" column
df_sorted = df.sort_values(by="Order")

# Print the DataFrame with all colors and aligned with order
print(df_sorted[["Color", "Value"]])
```
