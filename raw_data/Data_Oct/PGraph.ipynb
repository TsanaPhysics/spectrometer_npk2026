```python
import os
import pandas as pd
import matplotlib.pyplot as plt

# Specify the folder path
folder_path = "NdatasetCSV"

# List all CSV files in the folder starting with "BG_"
csv_files = [f for f in os.listdir(folder_path) if f.startswith("BG_") and f.endswith(".csv")]

# Create a subplot for each file
fig, axs = plt.subplots(len(csv_files), figsize=(8, 6 * len(csv_files)))

# Iterate through CSV files
for i, csv_file in enumerate(csv_files):
    # Construct the full file path
    file_path = os.path.join(folder_path, csv_file)

    # Load the CSV file
    df = pd.read_csv(file_path)

    # Extract 'No.' and 'Intensity' columns
    no = df['No.']
    intensity = df['Intensity']

    # Create a plot for each file
    axs[i].plot(no, intensity, label=csv_file)
    axs[i].set_xlabel('No.')
    axs[i].set_ylabel('Intensity')
    axs[i].set_title(f'Intensity vs. No. for {csv_file}')
    axs[i].legend()

# Adjust spacing between subplots
plt.tight_layout()

# Show the plot
plt.show()

```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt

# Specify the folder path
folder_path = "NdatasetCSV"

# List all CSV files in the folder starting with "BG_"
csv_files = [f for f in os.listdir(folder_path) if f.startswith("BG_") and f.endswith(".csv")]

# Create a DataFrame to store data from all "BG_" files
combined_df = pd.DataFrame()

# Iterate through the "BG_" files and plot their data
for csv_file in csv_files:
    # Construct the full file path
    file_path = os.path.join(folder_path, csv_file)

    # Load the CSV file
    df = pd.read_csv(file_path)

    # Plot the data for each "BG_" file
    plt.plot(df['No.'], df['Intensity'], label=csv_file)

    # Append the data to the combined DataFrame
    combined_df = pd.concat([combined_df, df], ignore_index=True)

# Set labels and title for the plot
plt.xlabel("No.")
plt.ylabel("Intensity")
plt.title("Intensity vs. No. for BG_ Files")

# Show legend for each "BG_" file
plt.legend()

# Show the plot
plt.show()

# If you want to save the combined data to a CSV file, you can use:
# combined_df.to_csv("combined_data.csv", index=False)

```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt

# Specify the folder path
folder_path = "PdatasetCSV"


# List of file names to include in the graph for N4
p2_file_names = ["P2_R.csv", "P2_A.csv", "P2_B.csv"]

# List of file names to include in the graph for N4
p4_file_names = ["P4_R.csv", "P4_A.csv", "P4_B.csv"]

# List of file names to include in the graph for N6
p6_file_names = ["P6_R.csv", "P6_A.csv", "P6_B.csv"]

# List of file names to include in the graph for N8
p8_file_names = ["P8_R.csv", "P8_A.csv", "P8_B.csv"]

# List of file names to include in the graph for N10
p10_file_names = ["P10_R.csv", "P10_A.csv", "P10_B.csv"]

# Function to create a plot for a list of file names
def create_plot(file_names, title):
    combined_df = pd.DataFrame()
    
    for csv_file in file_names:
        file_path = os.path.join(folder_path, csv_file)
        df = pd.read_csv(file_path)
        plt.plot(df['No.'], df['Intensity'], label=csv_file)
        combined_df = pd.concat([combined_df, df], ignore_index=True)
    
    plt.xlabel("No.")
    plt.ylabel("Intensity")
    plt.title(title)
    plt.legend()
    plt.show()
    
    # If you want to save the combined data to a CSV file, you can use:
    # combined_df.to_csv("combined_data.csv", index=False)

# Create plots for P2
create_plot(p2_file_names, "Intensity vs. No. for P2 Files")
    
# Create plots for P4
create_plot(p4_file_names, "Intensity vs. No. for P4 Files")

# Create plots for P6
create_plot(p6_file_names, "Intensity vs. No. for P6 Files")

# Create plots for P8
create_plot(p8_file_names, "Intensity vs. No. for P8 Files")

# Create plots for P10
create_plot(p10_file_names, "Intensity vs. No. for P10 Files")
```
