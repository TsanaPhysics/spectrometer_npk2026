```python
import pandas as pd
import matplotlib.pyplot as plt

# อ่านข้อมูลจากไฟล์ CSV และเก็บไว้ใน DataFrame
df = pd.read_csv('StandardData/PHT_Absorpt.csv')

# อ่านข้อมูลจากแถวแรกถึงแถวสุดท้าย
data = df.iloc[:, 1:]  # เริ่มต้นจากคอลัมน์ที่ 1 เนื่องจากคอลัมน์แรกเป็นชื่อข้อมูล

# สร้างกราฟเส้นแยกตามแต่ละเส้น
for i in range(len(data)):
    plt.plot(df.columns[1:], data.iloc[i], label=f'Concentration: {df["Concentrate"].iloc[i]}')

plt.xlabel('Colors')
plt.ylabel('Data')
plt.title('Line Graph for Different Concentrations')
plt.legend()
plt.grid(True)

# แสดงกราฟ
plt.show()
```
