```python
import os
import pandas as pd
import matplotlib.pyplot as plt

# Specify the folder path
folder_path = "PdatasetCSV"

# List all CSV files in the folder starting with "BG_"
csv_files = [f for f in os.listdir(folder_path) if f.startswith("BG_") and f.endswith(".csv")]

# Create a subplot for each file
fig, axs = plt.subplots(len(csv_files), figsize=(8, 6 * len(csv_files)))

# Iterate through CSV files
for i, csv_file in enumerate(csv_files):
    # Construct the full file path
    file_path = os.path.join(folder_path, csv_file)

    # Load the CSV file
    df = pd.read_csv(file_path)

    # Extract 'No.' and 'Intensity' columns
    no = df['No.']
    intensity = df['Intensity']

    # Create a plot for each file
    axs[i].plot(no, intensity, label=csv_file)
    axs[i].set_xlabel('No.')
    axs[i].set_ylabel('Intensity')
    axs[i].set_title(f'Intensity vs. No. for {csv_file}')
    axs[i].legend()

# Adjust spacing between subplots
plt.tight_layout()

# Show the plot
plt.show()

```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt

# Specify the folder path
folder_path = "PdatasetCSV"

# Define a function to generate the file names for each N value
def generate_file_names(n):
    return [f"P{n}_{suffix}.csv" for suffix in ["R", "A", "B"]]

# Generate file names for N4, N6, N8, and N10
p2_file_names = generate_file_names(2)
p4_file_names = generate_file_names(4)
p6_file_names = generate_file_names(6)
p8_file_names = generate_file_names(8)
p10_file_names = generate_file_names(10)



# Function to create a plot for a list of file names
def create_plot(file_names, title):
    combined_df = pd.DataFrame()
    
    for csv_file in file_names:
        file_path = os.path.join(folder_path, csv_file)
        df = pd.read_csv(file_path)
        plt.plot(df['No.'], df['Intensity'], label=csv_file)
        combined_df = pd.concat([combined_df, df], ignore_index=True)
    
    plt.xlabel("No.")
    plt.ylabel("Intensity")
    plt.title(title)
    plt.legend()
    plt.show()
    
    # If you want to save the combined data to a CSV file, you can use:
    # combined_df.to_csv("combined_data.csv", index=False)


# Create plots for P2
create_plot(p4_file_names, "Intensity vs. No. for P2 Files")
    
# Create plots for N4
create_plot(p4_file_names, "Intensity vs. No. for P4 Files")

# Create plots for N6
create_plot(p6_file_names, "Intensity vs. No. for P6 Files")

# Create plots for N8
create_plot(p8_file_names, "Intensity vs. No. for P8 Files")

# Create plots for N10
create_plot(p10_file_names, "Intensity vs. No. for P10 Files")
```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt

# Specify the folder path
folder_path = "PdatasetCSV"

# List of CSV files to be plotted
csv_files_to_plot = ["P2_blank.csv", "P2_W.csv", "P2_WR.csv"]

# Create an empty DataFrame to store data from selected files
combined_df = pd.DataFrame()

# Read and combine data from selected files
for csv_file in csv_files_to_plot:
    file_path = os.path.join(folder_path, csv_file)
    df = pd.read_csv(file_path)
    combined_df = pd.concat([combined_df, df['Intensity']], axis=1, ignore_index=True)

# Rename the columns to match the file names
combined_df.columns = csv_files_to_plot

# Plot the data
plt.figure(figsize=(10, 6))
for csv_file in csv_files_to_plot:
    plt.plot(combined_df.index, combined_df[csv_file], label=csv_file)

plt.xlabel("No.")
plt.ylabel("Intensity")
plt.title("Intensity vs. No.")
plt.legend()
plt.grid(True)

# Show the plot
plt.show()
```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt

# Specify the folder path
folder_path = "PdatasetCSV"

#


# List of lists containing CSV files to be plotted for each group
csv_files_to_plot_list = [
    [f"N{i}_{suffix}.csv" for suffix in ["R", "W", "NoR", "WR"]]
    for i in [2, 4, 6, 8, 10, 20, 40, 60, 80, 100]
]

# Plot the data for each group
for group_idx, csv_files_to_plot in enumerate(csv_files_to_plot_list):
    # Create an empty DataFrame to store data from selected files
    combined_df = pd.DataFrame()

    # Read and combine data from selected files
    for csv_file in csv_files_to_plot:
        file_path = os.path.join(folder_path, csv_file)
        df = pd.read_csv(file_path)
        combined_df = pd.concat([combined_df, df['Intensity']], axis=1, ignore_index=True)

    # Rename the columns to match the file names
    combined_df.columns = csv_files_to_plot

    # Plot the data
    plt.figure(figsize=(10, 6))
    for csv_file in csv_files_to_plot:
        plt.plot(combined_df.index, combined_df[csv_file], label=csv_file)

    plt.xlabel("No.")
    plt.ylabel("Intensity")
    plt.title(f"Intensity vs. No. (Group {group_idx + 1})")
    plt.legend()
    plt.grid(True)

    # Show the plot
    plt.show()
```
