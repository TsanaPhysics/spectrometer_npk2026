```python
import pandas as pd
import matplotlib.pyplot as plt

# ระบุตำแหน่งของไฟล์ CSV
file_path = "StandardData/P_Mean_L.csv"

# อ่านข้อมูลจากไฟล์ CSV
selected_data = pd.read_csv(file_path)

# ระบุคอลัมน์ที่ต้องการดึงข้อมูล
columns_to_display = ["Color"] + [col for col in selected_data.columns if col.startswith(("BG_0", "P2_B", "P4_B", "P6_B", "P8_B", "P10_B"))]

# แสดงข้อมูลที่เลือกบนหน้าจอ
selected_data = selected_data[columns_to_display]

# สร้างกราฟเส้น   KH2PO4 
plt.figure(figsize=(10, 6))
plt.plot(selected_data["Color"], selected_data["BG_0"], label="Back Ground")
plt.plot(selected_data["Color"], selected_data["P2_B"], label="2 mg/L")
plt.plot(selected_data["Color"], selected_data["P4_B"], label="4 mg/L")
plt.plot(selected_data["Color"], selected_data["P6_B"], label="6 mg/L")
plt.plot(selected_data["Color"], selected_data["P8_B"], label="8 mg/L")
plt.plot(selected_data["Color"], selected_data["P10_B"], label="10 mg/L")

# กำหนดแกน x
plt.xlabel("Color")

# กำหนดแกน y
plt.ylabel("Intensity")

# เพิ่มคำอธิบายกราฟ
plt.legend()

# แสดงกราฟ
plt.show()
```

```python
import pandas as pd

# ระบุตำแหน่งของไฟล์ CSV
file_path = "StandardData/P_Mean_L.csv"

# อ่านข้อมูลจากไฟล์ CSV
selected_data = pd.read_csv(file_path)

# ลบคอลัมน์ "BG_0" และเพิ่มข้อมูลจาก "BG_0" ไปยังคอลัมน์ "P2_B", "P4_B", "P6_B", "P8_B", "P10_B"

selected_data["P2_B"] = abs(selected_data["P2_B"] - selected_data["BG_0"])/selected_data["BG_0"]
selected_data["P4_B"] = abs(selected_data["P4_B"] - selected_data["BG_0"])/selected_data["BG_0"]
selected_data["P6_B"] = abs(selected_data["P6_B"] - selected_data["BG_0"])/selected_data["BG_0"]
selected_data["P8_B"] = abs(selected_data["P8_B"] - selected_data["BG_0"])/selected_data["BG_0"]
selected_data["P10_B"] = abs(selected_data["P10_B"] - selected_data["BG_0"])/selected_data["BG_0"]

# ระบุคอลัมน์ที่ต้องการดึงข้อมูล
columns_to_display = ["Color", "P2_B","P4_B", "P6_B", "P8_B", "P10_B"]

# แสดงข้อมูลที่เลือกบนหน้าจอ
selected_data = selected_data[columns_to_display]
print(selected_data)

# ระบุตำแหน่งและชื่อไฟล์ใหม่ที่ต้องการบันทึก
output_file_path = "StandardData/P_SelectL_Absorp.csv"

# บันทึกข้อมูลที่เลือกไว้ในไฟล์ CSV ใหม่
selected_data.to_csv(output_file_path, index=False)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# ระบุตำแหน่งของไฟล์ CSV
file_path = "StandardData/N_Select_Absorp.csv"

# อ่านข้อมูลจากไฟล์ CSV
selected_data = pd.read_csv(file_path)

# ระบุคอลัมน์ที่ต้องการดึงข้อมูล
columns_to_display = ["Color"] + [col for col in selected_data.columns if col.startswith(("BG_0", "N2_B", "N4_B", "N6_B", "N8_B", "N10_B"))]

# แสดงข้อมูลที่เลือกบนหน้าจอ
selected_data = selected_data[columns_to_display]

# สร้างกราฟแท่
plt.figure(figsize=(10, 5))
bar_width = 0.15  # กำหนดความกว้างของแท่

x = range(len(selected_data["Color"]))

plt.bar(x, selected_data["N2_B"], width=bar_width, label="N2_B")
plt.bar([i + bar_width for i in x], selected_data["N4_B"], width=bar_width, label="N4_B")
plt.bar([i + 2 * bar_width for i in x], selected_data["N6_B"], width=bar_width, label="N6_B")
plt.bar([i + 3 * bar_width for i in x], selected_data["N8_B"], width=bar_width, label="N8_B")
plt.bar([i + 4 * bar_width for i in x], selected_data["N10_B"], width=bar_width, label="N10_B")

# กำหนดแกน x
plt.xlabel("Color")

# กำหนดแกน y
plt.ylabel("Absorption")

# เพิ่มคำอธิบายกราฟ
plt.legend()

# แสดงกราฟ
plt.show()
```
