```python
import pandas as pd
import os

# Read the "N_mean.csv" file into a DataFrame
df = pd.read_csv("StandardData/N_StdDev.csv")

# Calculate the average of the columns following the "Color" column
# Group the data by the "Color" column and calculate the mean
average_by_color = df.groupby("Color").mean()

# Display the result
print("Average by Color:")
print(average_by_color)

# Define the output file path
output_folder = "StandardData"
output_filename = "N_StdDev.csv"
output_path = os.path.join(output_folder, output_filename)

# Check if the file "N_all_Mean.csv" already exists
if os.path.exists(output_path):
    print(f"'{output_filename}' already exists. Overwriting the file.")

# Save the DataFrame to a CSV file
average_by_color.to_csv(output_path)

print(f"Data saved to '{output_path}' as 'N_StdDev.csv'.")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# ระบุตำแหน่งของไฟล์ CSV
file_path = "StandardData/N_Select_AbsorpSD.csv"

# อ่านข้อมูลจากไฟล์ CSV
selected_data = pd.read_csv(file_path)

# เปลี่ยนชื่อคอลัมน์
selected_data = selected_data.rename(columns={"N2": "2.00 mg/L", "N4": "4.00 mg/L", "N6": "6.00 mg/L", "N8": "8.00 mg/L", "N10": "10.0 mg/L"})

# คอลัมน์ที่ต้องการแสดงในกราฟ
columns_to_display = ["2.00 mg/L", "4.00 mg/L", "6.00 mg/L", "8.00 mg/L", "10.0 mg/L"]
sd_columns = ["sd_N2", "sd_N4", "sd_N6", "sd_N8", "sd_N10"]
colors = selected_data["Color"].loc[(selected_data["Color"] != "White") & (selected_data["Color"] != "Black")]
data = selected_data[columns_to_display].loc[(selected_data["Color"] != "White") & (selected_data["Color"] != "Black")]
stddev_data = selected_data[sd_columns].loc[(selected_data["Color"] != "White") & (selected_data["Color"] != "Black")]

# สร้างกราฟแท่ง
plt.figure(figsize=(10, 6))

# สร้างกราฟแท่งสำหรับแต่ละคอลัมน์
bar_width = 0.15
index = range(len(colors))
cap_size = 7  # กำหนดขนาดขอบของ error bar

for i, (col, sd_col) in enumerate(zip(columns_to_display, sd_columns)):
    plt.bar(
        [x + i * bar_width for x in index],
        data[col],
        bar_width,
        yerr=stddev_data[sd_col],
        label=col,
        capsize=cap_size,  # กำหนดขนาดขอบของ error bar
        error_kw={"ecolor": "black", "capthick": 2, "elinewidth": 2, "markeredgewidth": 2}  # กำหนดลักษณะของ error bar เป็น T และ T กลับหัว
    )

# กำหนดแกน x
plt.xlabel("Color", fontsize=14, fontweight="bold")

# กำหนดแกน y
plt.ylabel("Percentage of Absorption", fontsize=14, fontweight="bold")

plt.yticks(fontsize=14, fontweight="bold")
plt.ylim(0.00, 100)  # กำหนดช่วงขอบเขตของแกน y

plt.xticks([x + (len(columns_to_display) / 2) * bar_width for x in index], colors, fontsize=12)
plt.yticks(fontsize=14, fontweight="bold")

# เพิ่มคำอธิบายกราฟ
plt.legend(fontsize=12, title="Concentration", title_fontsize=14)

# บันทึกภาพเป็นไฟล์ JPG ในโฟลเดอร์ "Images"
plt.savefig("Images/absorption_plot.jpg", format="jpg")

# แสดงกราฟ
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# ระบุตำแหน่งของไฟล์ CSV
file_path = "StandardData/N_Select_AbsorpSD.csv"

# อ่านข้อมูลจากไฟล์ CSV
selected_data = pd.read_csv(file_path)

# คอลัมน์ที่ต้องการแสดงในกราฟ
n_columns = ["N2", "N4", "N6", "N8", "N10"]
colors = selected_data["Color"].unique()
colors_mapping = {"Blue": "475nm", "Green": "520nm", "Red": "635nm"}
sd_columns = ["sd_" + col for col in n_columns]

# สร้างกราฟ
plt.figure(figsize=(10, 6))
# สร้างกราฟจุดแยกตามสี และแสดง error bar
for color in colors:
    if color not in ["Black", "White"]:
        data = selected_data[selected_data["Color"] == color]
        x = n_columns
        y = data[n_columns].values[0]  # เลือกข้อมูล N สำหรับสีนี้
        sd = data[sd_columns].values[0]  # เลือกข้อมูลความสัมพันธ์สำหรับสีนี้

        # แปลงสีเป็นความยาวคลื่น
        color_label = colors_mapping[color] if color in colors_mapping else color

        # กำหนดลักษณะของจุดตามสี
        marker = 'o'  # ตามค่าเริ่มต้น
        if color == "Red":
            marker = 'ro'  # จุดวงกลมสีแดง
        elif color == "Blue":
            marker = 'bs'  # สีเหลี่ยมสีน้ำเงิน
        elif color == "Green":
            marker = 'g^'  # ข้าวหลามตัดสีเขียว

        plt.errorbar(x, y, yerr=sd, fmt=marker, markersize=8, capsize=5, label=f"{color} : {colors_mapping[color]} ±5 nm")

        # เพิ่มเส้นกราฟ
        plt.plot(x, y, label="")  # เปลี่ยน label เป็นค่าว่าง

# กำหนดแกน x
plt.xlabel("Concentration (mg/L)", fontsize=16, fontweight="bold")

# กำหนดแกน y
plt.ylabel("Percentage of Absorption", fontsize=16, fontweight="bold")

# เพิ่มคำอธิบายกราฟ
legend_elements = [plt.Line2D([0], [0], marker='o', color='w', label=f"{color} : {colors_mapping[color]} ±5 nm", markerfacecolor=color, markersize=8) for color in colors if color not in ["Black", "White"]]
plt.legend(handles=legend_elements, fontsize=14, title="Color : Wave Length", title_fontsize=14)

# แก้ไขแกน x โดยกำหนดป้ายชื่อใหม่และปรับขนาดและน้ำหนักของตัวเลข
new_x_labels = ["2.00", "4.00", "6.00", "8.00", "10.0"]
plt.xticks(range(len(new_x_labels)), new_x_labels, fontsize=14, fontweight="bold")

# กำหนดขนาดแกน x และ y
plt.xticks(fontsize=14, fontweight="bold")
plt.yticks(fontsize=14, fontweight="bold")
plt.ylim(0.00, 100)  # กำหนดช่วงขอบเขตของแกน y

# บันทึกรูปภาพเป็นไฟล์ .jpg
plt.savefig("Images/absorption_linear_concentrate.jpg")

# แสดงกราฟ
plt.show()
```

```python
import pandas as pd
from scipy import stats

# Load your data
data = pd.read_csv("StandardData/N_Select_AbsorpSD.csv")

# Select the columns of interest
n_columns = ["N2", "N4", "N6", "N8", "N10"]
colors = ["Blue", "Green", "Red"]

# Perform linear regression for each color
for color in colors:
    color_data = data[data["Color"] == color]
    x = range(len(n_columns))
    y = color_data[n_columns].values[0]

    slope, intercept, r_value, p_value, std_err = stats.linregress(x, y)
    r_squared = r_value ** 2

    # Print the linear regression equation and R-squared value
    equation = f"y = {slope:.2f}x + {intercept:.2f}"
    print(f"Color: {color}")
    print(f"Linear Regression Equation: {equation}")
    print(f"R-squared: {r_squared:.2f}")
    print()
```

```python
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt

# Data
data = {
    "Color": ["Blue", "Green", "Red"],
    "Concentration": [2, 4, 6, 8, 10],
    "Percentage of Absorption": [
        [49.2176, 63.7268, 76.8980, 84.1028, 85.0981],
        [22.4398, 37.0154, 52.1301, 59.7957, 66.5390],
        [3.9377, 7.3214, 13.6193, 17.9969, 25.6989],
    ],
}

# Create a plot for each color
for color, y_data in zip(data["Color"], data["Percentage of Absorption"]):
    x_data = data["Concentration"]
    
    # Perform linear regression
    slope, intercept, r_value, p_value, std_err = stats.linregress(x_data, y_data)
    
    # Create the regression line
    x_range = np.linspace(min(x_data), max(x_data), 100)
    y_range = slope * x_range + intercept
    
    # Create the plot
    plt.figure(figsize=(10, 6))
    
    # Plot the data points
    plt.scatter(x_data, y_data, label=f"{color}", color=color.lower())
    
    # Plot the regression line
    plt.plot(x_range, y_range, label=f"y = {slope:.4f}x + {intercept:.4f}", linestyle='--')
    
    # Annotate with R²
    plt.annotate(f"R² = {r_value**2:.4f}", xy=(0.7, 0.1), xycoords='axes fraction', fontsize=12)
    
    # Set labels and title
    plt.xlabel("Concentration (mg/L)")
    plt.ylabel("Percentage of Absorption")
    plt.title(f"Absorption for {color}")
    plt.legend()
    
    # Save the plot as a .png file
    plt.savefig(f"{color}_absorption_graph.png")
    
    # Show the plot
    plt.show()
```
