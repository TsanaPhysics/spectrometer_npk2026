```python
import pandas as pd

# ระบุตำแหน่งของไฟล์ CSV
file_path = "StandardData/N_Mean_Selected.csv"

# อ่านข้อมูลจากไฟล์ CSV
selected_data = pd.read_csv(file_path)

# ลบคอลัมน์ "BG_0" และเพิ่มข้อมูลจาก "BG_0" ไปยังคอลัมน์ "N2_B", "N4_B", "N6_B", "N8_B", "N10_B"
selected_data["N2_B"] = abs(selected_data["N2_B"] - selected_data["BG_0"])/selected_data["BG_0"]
selected_data["N4_B"] = abs(selected_data["N4_B"] - selected_data["BG_0"])/selected_data["BG_0"]
selected_data["N6_B"] = abs(selected_data["N6_B"] - selected_data["BG_0"])/selected_data["BG_0"]
selected_data["N8_B"] = abs(selected_data["N8_B"] - selected_data["BG_0"])/selected_data["BG_0"]
selected_data["N10_B"] = abs(selected_data["N10_B"] - selected_data["BG_0"])/selected_data["BG_0"]

# ระบุคอลัมน์ที่ต้องการดึงข้อมูล
columns_to_display = ["Color", "N2_B", "N4_B", "N6_B", "N8_B", "N10_B"]

# แสดงข้อมูลที่เลือกบนหน้าจอ
selected_data = selected_data[columns_to_display]
print(selected_data)

# ระบุตำแหน่งและชื่อไฟล์ใหม่ที่ต้องการบันทึก
output_file_path = "StandardData/N_Select_Absorp.csv"

# บันทึกข้อมูลที่เลือกไว้ในไฟล์ CSV ใหม่
selected_data.to_csv(output_file_path, index=False)
```
