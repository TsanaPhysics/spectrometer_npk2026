```python
import csv
import pandas as pd

# เปิดไฟล์ CSV ด้วยฟังก์ชัน open
with open('StandardData/NHT_Absorpt.csv', newline='') as csvfile:
    # ใช้ csv.reader เพื่ออ่านข้อมูลจากไฟล์ CSV
    csv_reader = csv.reader(csvfile)

    # สร้างชื่อไฟล์ CSV ที่ต้องการบันทึกในโฟลเดอร์ ATSSE2023
    output_file = 'ATSSE2023/NHT_Absorpt_modified.csv'

    # เปิดไฟล์ CSV สำหรับบันทึก
    with open(output_file, 'w', newline='') as output_csvfile:
        # ใช้ csv.writer เพื่อเขียนข้อมูลลงในไฟล์ CSV ใหม่
        csv_writer = csv.writer(output_csvfile)

        for row in csv_reader:
            # ตรวจสอบและแทนค่าตามที่คุณต้องการ
            modified_row = [value if value not in ['Blue', 'Green', 'Red'] else
                            '470' if value == 'Blue' else
                            '520' if value == 'Green' else
                            '635' for value in row]

            # เขียนแถวที่แก้ไขแล้วลงในไฟล์ CSV ใหม่
            csv_writer.writerow(modified_row)

print(f"ไฟล์ CSV แก้ไขและบันทึกใน {output_file}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the CSV data into a pandas DataFrame
data = pd.read_csv('StandardData/NHT_Absorpt.csv')

# Extract the data for the 'Blue', 'Green', 'Red', and 'White' columns
colors = ['Blue', 'Green', 'Red', 'White']

# Create a horizontal bar plot for each row in the DataFrame
for index, row in data.iterrows():
    plt.barh(colors, row[colors], label=f'Row {index+1}')

# Set labels and title
plt.xlabel('Values')
plt.ylabel('Colors')
plt.title('Color Data from CSV')

# Show a legend
plt.legend()

# Display the plot
plt.show()
```
