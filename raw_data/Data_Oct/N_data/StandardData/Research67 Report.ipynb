```python
from sklearn.linear_model import LinearRegression
import numpy as np

# ข้อมูล x
x = np.array([2, 4, 6, 8, 10]).reshape(-1, 1)

# ข้อมูล y
y = np.array([95.41, 95.27, 95.18, 95.04, 95.53])

# สร้าง Linear Regression model
model = LinearRegression()

# Fit model กับข้อมูล
model.fit(x, y)

# หาค่าความชัน (slope) และค่าตัดแกน y (y-intercept)
m = model.coef_[0]
b = model.intercept_

print("ค่าความชัน (slope):", m)
print("ค่าตัดแกน y (y-intercept):", b)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# ข้อมูล C และ G
C = np.array([2, 4, 6, 8, 10]).reshape(-1, 1)
G = np.array([95.41, 95.27, 95.18, 95.04, 94.93])

# สร้างโมเดล Linear Regression
model = LinearRegression().fit(C, G)

# ทำนาย G จาก C
G_pred = model.predict(C)

# คำนวณค่า R^2
r_squared = r2_score(G, G_pred)

# พล็อตกราฟ
plt.scatter(C, G, color='blue', label='Actual Data')
plt.plot(C, G_pred, color='red', label='Regression Line')
plt.title('Linear Regression')
plt.xlabel('C')
plt.ylabel('G')
plt.legend()
plt.show()

# แสดงสมการ
print("สมการทดถอย: G =", model.intercept_, "+", model.coef_[0], "* C")
print("ค่า R^2:", r_squared)
```

```python
import os
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
import pandas as pd

# สร้างโฟลเดอร์ 'img_research66' ถ้ายังไม่มีอยู่
os.makedirs('img_research66', exist_ok=True)

# ข้อมูล NW Concentration และ WB Reflectance
NW_Concentration = np.array([2, 4, 6, 8, 10]).reshape(-1, 1)
WB_Reflectance = np.array([94.7615, 94.7268, 94.6963, 94.675, 94.6606])
SD = np.array([4.5300, 4.5100, 4.5700, 4.9150, 4.5150])

# สร้างโมเดล Linear Regression
model = LinearRegression().fit(NW_Concentration, WB_Reflectance)

# ทำนาย WB Reflectance จาก NW Concentration
WB_Reflectance_pred = model.predict(NW_Concentration)

# คำนวณค่า R^2
r_squared = r2_score(WB_Reflectance, WB_Reflectance_pred)

# พล็อตกราฟ
plt.errorbar(NW_Concentration, WB_Reflectance, yerr=SD, fmt='o', color='blue', uplims=True, lolims=True, label='Actual Data')
plt.plot(NW_Concentration, WB_Reflectance_pred, color='red', label='Regression Line')
plt.title('Linear Regression')
plt.xlabel('Nitrogen Concentration (%)')
plt.ylabel('%Blue Color Reflectance (%) : WHITE Source')
plt.legend()

# แสดงสมการและ R^2 บนกราฟ
plt.text(2, 94.66, f'%WB_Reflectance = {model.coef_[0]:.4f} * N_Concentration + {model.intercept_:.4f}', fontsize=10)
plt.text(2, 94.65, f'R^2 = {r_squared:.4f}', fontsize=10)

# บันทึกภาพกราฟ
plt.savefig('img_research66/WB_Reflectance_vs_NW_Concentration.jpg')

# สร้าง DataFrame จากข้อมูล
data = {
    'NW Concentration': NW_Concentration.flatten(),
    'WB Reflectance': WB_Reflectance,
    'SD': SD
}
df = pd.DataFrame(data)

# บันทึกเป็นไฟล์ .csv
df.to_csv('img_research66/research_data_NW.csv', index=False)

plt.show()
```
