```python
import pandas as pd

# ระบุตำแหน่งของไฟล์ CSV
file_path = "StandardData/NAsorpt.csv"

# อ่านข้อมูลจากไฟล์ CSV
data = pd.read_csv(file_path)

# สร้างรายการของคอลัมน์ที่คุณต้องการคำนวณ
columns_to_calculate = [
    ("N2_WR", "N2_B"),
    ("N4_WR", "N4_B"),
    ("N6_WR", "N6_B"),
    ("N8_WR", "N8_B"),
    ("N10_WR", "N10_B"),
]

# สร้าง DataFrame ที่มีเฉพาะคอลัมน์ "Color"
result = data[["Color"]]

# คำนวณและเพิ่มคอลัมน์ของผลต่างลงใน DataFrame
for col1, col2 in columns_to_calculate:
    result[f"{col1.replace('N', '').replace('_WR', '')} mg/L"] = data[col1] - data[col2]

# เปลี่ยนหัวตารางให้แสดงหน่วยในรูปแบบที่คุณต้องการ
column_mapping = {
    "N2 mg/L": "Difference_N2_WR_N2_B",
    "N4 mg/L": "Difference_N4_WR_N4_B",
    "N6 mg/L": "Difference_N6_WR_N6_B",
    "N8 mg/L": "Difference_N8_WR_N8_B",
    "N10 mg/L": "Difference_N10_WR_N10_B",
}

result = result.rename(columns=column_mapping)

# แสดงผลตาราง
print(result)
```

```python
import pandas as pd

# โหลดไฟล์ NAsorpt.csv
data = pd.read_csv('StandardData/NAsorpt.csv')

# ข้อมูลในแถวที่ 0
row_0 = data.iloc[0, 1:7]

# สร้าง DataFrame เพื่อเก็บผลต่างข้อมูลของแต่ละแถว
absorptance = pd.DataFrame()

# วนลูปผ่านแถวที่ 1 ถึงแถวสุดท้ายของไฟล์
for i in range(1, len(data)):
    # ข้อมูลในแถวปัจจุบัน
    current_row = data.iloc[i, 1:7]

    # คำนวณผลต่าง
    absorpt = (current_row - row_0) * 100 / current_row

    # เพิ่มผลต่างข้อมูลใน DataFrame
    absorptance[f'Row {i}'] = absorpt

# แสดงผลลัพธ์
print(absorptance)

```

```python
import pandas as pd

file_path = "StandardData/N_all_Mean.csv"
data = pd.read_csv(file_path)

columns = ["Color"] + [f"N{num}_{ext}" for num in [2, 4, 6, 8, 10] for ext in ["B", "WR"]]

selected_data = data[columns]

for num in [2, 4, 6, 8, 10]:
    for ext in ["B", "WR"]:
        col_name = f"N{num}_{ext}"
        base_col_name = f"N{num}_B"
        selected_data[f"{col_name}_B"] = (abs(selected_data[col_name] - selected_data[base_col_name]) * 100) / selected_data[col_name]

print(selected_data)

```
